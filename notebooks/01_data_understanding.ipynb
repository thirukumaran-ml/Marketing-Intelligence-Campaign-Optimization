```python
numeric_kpis = [
    "impressions",
    "clicks",
    "conversions",
    "ad_spend",
    "revenue",
    "CTR",
    "conversion_rate",
    "CPC",
    "CPA",
    "ROAS",
    "profit"
]

outlier_summary = []

for column in numeric_kpis:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_summary.append({
        "Variable": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": outliers,
        "Outlier_Percentage": (outliers / len(df)) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    "Outlier_Percentage",
    ascending=False
)
```

# Phase 1 — Data Understanding: Conclusions

## 1. Dataset Overview

The advertising campaign dataset contains **10,000 observations and 41 variables**. The dataset consists of **20 numerical variables and 21 categorical/non-numerical variables**.

The variables cover multiple aspects of digital advertising campaigns, including campaign characteristics, platforms, creative attributes, audience characteristics, campaign quality and cost, performance metrics, financial outcomes, user engagement, and industry information.

---

## 2. Data Quality Assessment

The initial data-quality assessment produced the following results:

- **Total observations:** 10,000
- **Total variables:** 41
- **Missing values:** 0
- **Duplicate records:** 0
- **Numerical variables:** 20
- **Categorical/non-numerical variables:** 21

Therefore, the dataset contains no missing values or duplicate records and is structurally suitable for further analysis.

Logical consistency checks also showed:

- No cases where clicks exceeded impressions.
- No cases where conversions exceeded clicks.
- No negative impressions.
- No negative clicks.
- No negative conversions.
- No negative advertising spend.
- No negative revenue.

These results indicate that the major campaign volume and financial variables satisfy the basic logical constraints expected from advertising data.

## 4. Target Variables

The project focuses on four major campaign performance targets:

1. **CTR (Click-Through Rate)**
2. **Conversion Rate**
3. **ROAS (Return on Ad Spend)**
4. **Profit**

These targets will be used in the predictive modeling stage to estimate campaign performance and support marketing optimization decisions.

---

## 5. KPI Validation

The major campaign KPIs were validated against their underlying campaign metrics.

The following relationships were confirmed:

- **CTR = (Clicks / Impressions) × 100**
- **Conversion Rate = (Conversions / Clicks) × 100**
- **ROAS = Revenue / Ad Spend**
- **Profit = Revenue − Ad Spend**

The differences between the stored KPI values and the calculated values were extremely small, indicating that the stored KPI values are consistent with their underlying metrics, with minor differences attributable to rounding.

CPC and Actual CPC were also found to be derived from advertising spend and clicks.

---

## 6. Zero-Conversion Consideration

The dataset contains campaigns with zero conversions. Therefore, metrics such as CPA, which depend on the number of conversions, require special consideration.

For campaigns with zero conversions:

**CPA = Ad Spend / Conversions**

cannot be calculated using ordinary division because the denominator is zero.

This issue will be handled appropriately during the data preprocessing and feature engineering stages.

## 9. Predictive Modeling Considerations

The KPI validation revealed an important modeling consideration.

Since CTR, Conversion Rate, ROAS, and Profit are mathematically derived from other campaign metrics, directly using their underlying components as predictors could result in **target leakage or trivial prediction relationships**.

For example:

- Using clicks and impressions to predict CTR could allow the model to reproduce the CTR formula.
- Using conversions and clicks to predict Conversion Rate could produce a similar issue.
- Using revenue and ad spend to predict ROAS would directly expose the components of the target.
- Using revenue and ad spend to predict Profit would directly expose the profit calculation.

Therefore, predictive models should primarily focus on campaign characteristics, audience attributes, platform information, creative attributes, quality-related variables, and other information that can reasonably be available independently of the final campaign outcome.

The final feature-selection strategy will be determined during **Feature Engineering and Predictive Modeling**.
